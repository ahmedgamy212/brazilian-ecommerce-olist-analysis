```python
date_cols = [
    'order_purchase_timestamp',
    'order_approved_at',
    'order_delivered_carrier_date',
    'order_delivered_customer_date',
    'order_estimated_delivery_date',
]

for c in date_cols:
    orders[c] = pd.to_datetime(orders[c])
```

```python
mask = (orders['order_status'] == 'delivered') & orders['order_approved_at'].isna()

print(mask.sum())

orders.loc[mask, 'order_approved_at'] = orders.loc[mask, 'order_purchase_timestamp']

ok = (
    (orders['order_status'] == 'delivered') &
    orders['order_approved_at'].notna() &
    orders['order_delivered_carrier_date'].notna()
)

diff = orders.loc[ok, 'order_delivered_carrier_date'] - orders.loc[ok, 'order_approved_at']
lag = diff.median()
print(lag)


mask = (
    (orders['order_status'] == 'delivered') &
    orders['order_delivered_carrier_date'].isna()
)
print(mask.sum()) 

orders.loc[mask, 'order_delivered_carrier_date'] = (
    orders.loc[mask, 'order_approved_at'] + lag
)
ok = (
    (orders['order_status'] == 'delivered') &
    orders['order_delivered_carrier_date'].notna() &
    orders['order_delivered_customer_date'].notna()
)

diff = orders.loc[ok, 'order_delivered_customer_date'] - orders.loc[ok, 'order_delivered_carrier_date']
lag = diff.median()
print(lag)

mask = (
    (orders['order_status'] == 'delivered') &
    orders['order_delivered_customer_date'].isna()
)
print(mask.sum())  

orders.loc[mask, 'order_delivered_customer_date'] = (
    orders.loc[mask, 'order_delivered_carrier_date'] + lag
)

orders['order_delivered_customer_date'].isna().sum() 
```
