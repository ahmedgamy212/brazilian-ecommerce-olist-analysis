```python
mapping = dict(zip(store_id_relation2['hpg_store_id'], store_id_relation2['air_store_id']))
hpg_reserve2['air_store_id'] = hpg_reserve2['hpg_store_id'].map(mapping)
hpg_store_info2['air_store_id'] = hpg_store_info2['hpg_store_id'].map(mapping) 
```

```python
base = air_visit_data.copy()
base['visit_date'] = pd.to_datetime(base['visit_date'])

full = []
for sid, g in base.groupby('air_store_id'):
    idx = pd.date_range(g['visit_date'].min(), g['visit_date'].max(), freq='D')
    g = g.set_index('visit_date').reindex(idx)
    g['air_store_id'] = sid
    g.index.name = 'visit_date'
    full.append(g.reset_index())
base = pd.concat(full, ignore_index=True)
```

```python
cols = ['air_store_id', 'visit_datetime', 'reserve_datetime', 'reserve_visitors']
res = pd.concat([air_reserve[cols], hpg_reserve2[cols]], ignore_index=True)
res['visit_date'] = pd.to_datetime(res['visit_datetime']).dt.normalize()
res = (res.groupby(['air_store_id', 'visit_date'], as_index=False)['reserve_visitors'].sum())
```

```python
di = date_info.rename(columns={'calendar_date': 'visit_date'})
di['visit_date'] = pd.to_datetime(di['visit_date'])
```

```python
store_cols = ['air_store_id', 'air_genre_name', 'air_area_name', 'latitude', 'longitude']
train = (base
         .merge(res, on=['air_store_id', 'visit_date'], how='left', validate='1:1')
         .merge(air_store_info[store_cols], on='air_store_id', how='left', validate='m:1')
         .merge(di, on='visit_date', how='left', validate='m:1'))
train['reserve_visitors'] = train['reserve_visitors'].fillna(0)
```

```python
def build_features(df):
    df = df.copy()
    df = df.sort_values(['air_store_id', 'visit_date']).reset_index(drop=True)
 
    d = df['visit_date']
    df['year']           = d.dt.year
    df['month']          = d.dt.month
    df['day']            = d.dt.day
    df['weekofyear']     = d.dt.isocalendar().week.astype(int)
    df['dayofyear']      = d.dt.dayofyear
    df['quarter']        = d.dt.quarter
    df['is_weekend']     = df['day_of_week'].isin(['Saturday', 'Sunday']).astype(int)
    df['is_month_start'] = d.dt.is_month_start.astype(int)
    df['is_month_end']   = d.dt.is_month_end.astype(int)
 
    g = df.groupby('air_store_id')['visitors']
    for lag in [1, 2, 3, 7, 14, 28, 364]:
        df[f'lag_{lag}'] = g.shift(lag)
 
    for window in [7, 14, 28, 56]:
        df[f'rolling_mean_{window}'] = g.transform(
            lambda s: s.shift(1).rolling(window, min_periods=1).mean())
        df[f'rolling_std_{window}'] = g.transform(
            lambda s: s.shift(1).rolling(window, min_periods=1).std())
 
    df['expanding_mean'] = g.transform(lambda s: s.shift(1).expanding().mean())
    df['lag_7_diff_lag_14'] = df['lag_7'] - df['lag_14']
 
    r = df.groupby('air_store_id')['reserve_visitors']
    for lag in [1, 7]:
        df[f'reserve_lag_{lag}'] = r.shift(lag)
    df['reserve_rolling_mean_7'] = r.transform(
        lambda s: s.shift(1).rolling(7, min_periods=1).mean())
 
    return df
 
df = build_features(train)
 
df = df.dropna(subset=['visitors']).reset_index(drop=True)
df['log_visitors'] = np.log1p(df['visitors'])
print(f"Shape after FE: {df.shape}")
 
X_original = df.drop(columns=['visitors', 'log_visitors']).copy()
y_original = df['log_visitors'].copy()
 
cat_features = ['air_genre_name', 'air_area_name', 'day_of_week']
```

```python
def prepare_data(X_raw, y_raw, for_catboost=False):
    X = X_raw.copy()
    y = y_raw.copy()
 
    order = X['visit_date'].sort_values(kind='stable').index
    X, y = X.loc[order], y.loc[order]
    dates = X['visit_date']
 
    cutoff = dates.iloc[int(len(dates) * 0.8)]
    is_train = (dates < cutoff).values
 
    X = X.drop(columns=['air_store_id', 'visit_date'], errors='ignore')
 
    for col in cat_features:
        if for_catboost:
            X[col] = X[col].fillna('unknown').astype(str)
        else:
            X[col] = X[col].fillna('unknown').astype('category')
 
    X_train, X_test = X[is_train], X[~is_train]
    y_train, y_test = y[is_train], y[~is_train]
    return X_train, X_test, y_train, y_test
```

```python

Xtr, Xte, ytr, yte = prepare_data(X_original, y_original, for_catboost=True)

cb_param_grid = {
    'iterations': [500, 1000],
    'learning_rate': [0.03, 0.05],
    'depth': [6, 8],
    'l2_leaf_reg': [3, 5],
}

cb_search = RandomizedSearchCV(
    CatBoostRegressor(loss_function='RMSE', random_seed=42, verbose=0),
    param_distributions=cb_param_grid,
    n_iter=10,
    cv=TimeSeriesSplit(n_splits=3),
    scoring='neg_root_mean_squared_error',
    n_jobs=1,
    verbose=1,
    error_score='raise',
    random_state=42,
)
cb_search.fit(Xtr, ytr, cat_features=cat_features)
cb_best = cb_search.best_estimator_
joblib.dump(cb_best, 'catboost_best_model.pkl')

y_pred_cb = np.expm1(cb_best.predict(Xte))
y_true = np.expm1(yte)
cb_rmse = np.sqrt(mean_squared_error(y_true, y_pred_cb))
cb_mae  = mean_absolute_error(y_true, y_pred_cb)
cb_r2   = r2_score(y_true, y_pred_cb)
print(f"CatBoost → RMSE: {cb_rmse:.4f} | MAE: {cb_mae:.4f} | R²: {cb_r2:.4f}")
```

```python

Xtr, Xte, ytr, yte = prepare_data(X_original, y_original, for_catboost=False)

lgb_param_grid = {
    'n_estimators': [500, 1000],
    'learning_rate': [0.03, 0.05],
    'max_depth': [5, 7, 9],
    'num_leaves': [31, 50],
    'min_child_samples': [20, 50],
}

lgb_search = RandomizedSearchCV(
    LGBMRegressor(random_state=42, n_jobs=-1, verbose=-1),
    param_distributions=lgb_param_grid,
    n_iter=10,
    cv=TimeSeriesSplit(n_splits=3),
    scoring='neg_root_mean_squared_error',
    n_jobs=1,
    verbose=1,
    random_state=42,
    error_score='raise',
)
lgb_search.fit(Xtr, ytr, categorical_feature=cat_features)
lgb_best = lgb_search.best_estimator_
joblib.dump(lgb_best, 'lgb_best_model.pkl')

y_pred_lgb = np.expm1(lgb_best.predict(Xte))
lgb_rmse = np.sqrt(mean_squared_error(y_true, y_pred_lgb))
lgb_mae  = mean_absolute_error(y_true, y_pred_lgb)
lgb_r2   = r2_score(y_true, y_pred_lgb)
print(f"LightGBM → RMSE: {lgb_rmse:.4f} | MAE: {lgb_mae:.4f} | R²: {lgb_r2:.4f}")
```

```python
Xtr, Xte, ytr, yte = prepare_data(X_original, y_original, for_catboost=False)

xgb_param_grid = {
    'n_estimators': [500, 1000],
    'learning_rate': [0.03, 0.05],
    'max_depth': [5, 7, 9],
    'min_child_weight': [1, 5],
    'subsample': [0.8, 1.0],
    'colsample_bytree': [0.8, 1.0],
}

xgb_search = RandomizedSearchCV(
    XGBRegressor(
        objective='reg:squarederror',
        tree_method='hist',
        enable_categorical=True,
        random_state=42,
        n_jobs=-1,
        verbosity=0,
    ),
    param_distributions=xgb_param_grid,
    n_iter=10,
    cv=TimeSeriesSplit(n_splits=3),
    scoring='neg_root_mean_squared_error',
    n_jobs=1,
    verbose=1,
    random_state=42,
    error_score='raise',
)
xgb_search.fit(Xtr, ytr)
xgb_best = xgb_search.best_estimator_
joblib.dump(xgb_best, 'xgb_best_model.pkl')

y_pred_xgb = np.expm1(xgb_best.predict(Xte))
xgb_rmse = np.sqrt(mean_squared_error(y_true, y_pred_xgb))
xgb_mae  = mean_absolute_error(y_true, y_pred_xgb)
xgb_r2   = r2_score(y_true, y_pred_xgb)
print(f"XGBoost → RMSE: {xgb_rmse:.4f} | MAE: {xgb_mae:.4f} | R²: {xgb_r2:.4f}")
```

```python
comparison = pd.DataFrame({
    'Model':       ['CatBoost', 'LightGBM', 'XGBoost'],
    'RMSE':        [cb_rmse, lgb_rmse, xgb_rmse],
    'MAE':         [cb_mae,  lgb_mae,  xgb_mae],
    'R2':          [cb_r2,   lgb_r2,   xgb_r2],
    'Best_Params': [
        str(cb_search.best_params_),
        str(lgb_search.best_params_),
        str(xgb_search.best_params_),
    ],
})
print("\n" + "="*60)
print("📊 Final Comparison")
print("="*60)
print(comparison.to_string(index=False))
```
